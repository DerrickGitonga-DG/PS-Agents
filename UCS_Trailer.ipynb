```python
import heapq

# The graph from Part 1
graph = {
    "Nairobi": {"Rungiri": 19},
    "Rungiri": {"Gitaru": 4.1},
    "Gitaru": {"Limuru": 15},
    "Limuru": {"Kimende Township": 14.5},
    "Kimende Township": {"Gilgil": 68.4},
    "Gilgil": {"Nakuru East": 30},
    "Nakuru East": {"Salga": 37},
    "Salga": {"Kericho": 78},
    "Kericho": {"Kapsoit": 11},
    "Kapsoit": {"Kaitui": 11},
    "Kaitui": {"Awasi": 20},
    "Awasi": {"Ahero": 18},
    "Ahero": {"Orongo": 15},
    "Orongo": {"Kisumu": 11},
    "Kisumu": {},
}

def ucs(graph, start, goal):
    frontier = [(0, start, [start])]   # (path cost g, node, path so far)
    explored = set()
    step = 1

    while frontier:
        cost, node, path = heapq.heappop(frontier)   # cheapest node first
        print(f"Step {step}: expand {node} (g = {round(cost, 1)} km)")

        if node == goal:
            print("  Goal reached!")
            return path, cost

        if node in explored:
            continue
        explored.add(node)

        for neighbour, dist in graph.get(node, {}).items():
            if neighbour not in explored:
                heapq.heappush(frontier, (cost + dist, neighbour, path + [neighbour]))

        print("  Frontier:", [(n, round(c, 1)) for c, n, _ in sorted(frontier)])
        print("  Explored:", sorted(explored))
        print()
        step += 1

    return None, float("inf")

path, cost = ucs(graph, "Nairobi", "Kisumu")

print("\nOptimal path:", " -> ".join(path))
print("Total cost:", round(cost, 1), "km")
```
